## Three construct candidates

### Candidate 1 — broad

**Any temporal change in one or more typed defensive relationships.**

**Rejected.** Nearly every active-play window satisfies it. It cannot distinguish 1888 coordinated translation, 4197 heterogeneous activity, or ordinary continuous motion from a meaningful reconfiguration.

### Candidate 2 — cross-scale

**A temporally bounded change involving multiple defensive relational scales.**

**Rejected as a requirement and insufficient alone.** Generic activity and ball/shared translation can move several panels together; tackle is entangled; 1230 is membership-sensitive; 590 provides meaningful focal departure without robust local agreement. Requiring multiple scales would discard supported focal-scale behavior and revive an unsupported propagation theory.

### Candidate 3 — reference-relative geometric reconfiguration

**A coherent, temporally localized change in prospectively specified typed defensive relationships that is not adequately described by the relevant pre-specified baseline motion alone.**

**Retained provisionally as a conceptual candidate, not yet operationalized.** “Coherent,” “localized,” “adequately described,” and “relevant baseline” still require prospective validation. They cannot become hidden analyst-selected thresholds or references. Candidate 3 is narrower than generic movement, permits one or several scales, and does not require a motif, state, responsibility, or propagation.

## Adversarial tests

### Generic movement

“Movement occurred” but “reconfiguration did not” when a pre-specified collective/ball baseline accounts for the dominant geometry and residual typed relationships show no prospectively coherent departure. Thus 1888 is mostly coordinated movement, not a necessary reallocation event. The 4197 case contains substantial relational activity/deformation without coherent reference-relative organization. The 590 focal departure is a stronger candidate because focal leave-one-out change is prominent relative to collective movement, although opponent interpretation remains sensitive.

This distinction is not yet an automated rule. If matched ordinary movement cannot be separated prospectively, the construct fails.

### Reference dependence

The construct is inherently reference-relative. That is scientifically acceptable only when the reference and membership rule are specified before the outcome window and justified for the geometric question. Phase 2E–2G show that post-hoc reference selection is unacceptable: it can reverse 1232 and rescue 1230 compression. Repeated contradictions among reasonable prospective references would falsify stability.

### Scale

Reconfiguration may occur within one scale or across scales. Collective translation alone is normally baseline movement; focal departure may be meaningful without robust local change; local deformation is insufficient because it is common; cross-scale agreement is a possible indicator, not a requirement.

### Non-zero-sum relationships

No definition may imply that relational strength or “attention” transfers from a fixed budget. Opponent responsiveness and collective coherence may increase together. “Reallocation” is therefore risky language.

### Cognition

Tracking supports behavior and geometry only. The construct must not imply attention, decision, responsibility, recognition, instruction, intent, or mental workload. These restrictions favor **relational reconfiguration**, **defensive geometric reconfiguration**, or **reference-relative defensive reorganization** in empirical writing.

## Terminology decision: B

**Retain “defensive relational reallocation” provisionally as historical/theoretical shorthand, but prefer “relational reconfiguration” in empirical sections.**

“Reallocation” conveniently names a changing relational balance, but it risks latent attention/responsibility-transfer and zero-sum interpretations that tracking cannot support. “Relational reconfiguration” better describes observable geometry while preserving the project’s historical development. The project is not renamed automatically.

## Minimum defensible definition

> **Defensive relational reconfiguration is a coherent, temporally localized change in prospectively specified typed defensive geometric relationships that is not adequately described by the relevant pre-specified baseline motion alone.**

This is a provisional conceptual definition, not an operational detector. It excludes generic collective translation, generic local deformation, and arbitrary active-play movement only if “coherent,” “localized,” and “adequately described” can be validated prospectively against contrasts. If they cannot, the definition—and the construct—must be retired.

## Necessary conceptual conditions versus possible indicators

### Necessary conceptual conditions

1. Relationships and baseline reference are typed, observable, and prospectively specified.
2. Change is temporally localized enough to distinguish it from the surrounding continuous movement.
3. The pattern contains coherent reference-relative change not adequately represented by baseline motion alone.
4. The interpretation remains behavioral/geometric, non-zero-sum, non-cognitive, non-causal, and does not require responsibility.
5. The conclusion is robust enough to reasonable pre-declared reference sensitivity that it is not a post-hoc selection artifact.

### Possible observable indicators—not individually necessary or sufficient

- focal departure from collective motion;
- fixed local-configuration deformation;
- prospectively selected opponent-relative change;
- cross-scale correspondence or disagreement;
- persistence beyond isolated tracking noise;
- broad ball/collective residual structure.

No indicator is promoted to a score, threshold, or required motif.

## Explicit future falsification criteria

The construct should be rejected or materially narrowed if:

1. Prospectively defined references cannot distinguish hypothesized reconfiguration from matched ordinary movement.
2. Apparent effects vanish or reverse across additional matches or data sources.
3. Reasonable pre-specified reference choices produce contradictory conclusions too often.
4. Ball and collective motion adequately explain all apparent focal/local/opponent cases.
5. Independent expert visual interpretation fails to correspond to the measured typed geometry.
6. Operationalization requires post-hoc relationship, threshold, window, or reference tuning.
7. Generic active play satisfies the necessary conditions at a rate that makes the construct non-discriminating.
8. “Coherence” and “adequately described” cannot be defined prospectively without simply encoding the desired examples.

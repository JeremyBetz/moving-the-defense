# Phase 3A: prospective matched-contrast validation design

## Design-only scope

This notebook freezes an internal geometric validation protocol. It does **not** enumerate candidates, sample controls, compute relational outcomes, inspect separation, or execute Phase 3B. Event anchors are metric-independent sampling mechanisms, never ground-truth labels.

> Defensive relational reconfiguration is a coherent, temporally localized change in prospectively specified typed defensive geometric relationships that is not adequately described by the relevant pre-specified baseline motion alone.

The strongest possible future result is internal geometric discrimination/robustness, not tactical or semantic truth.

```python
from pathlib import Path
import json, pandas as pd
from IPython.display import display
ROOT=Path.cwd().resolve(); ROOT=ROOT.parent if ROOT.name=='notebooks' else ROOT
cfg=json.loads((ROOT/'config'/'phase3a_validation_protocol.json').read_text())
assert cfg['status']=='FROZEN_DESIGN_NOT_EXECUTED'
print({'version':cfg['phase3a_protocol_version'],'status':cfg['status'],'seed':cfg['random_seed'],'dataset':cfg['dataset'],'primary_window':cfg['window']['primary_seconds'],'sensitivity_window':cfg['window']['sensitivity_seconds']})
```

## Anchor decision

Open-play PASS events with nonmissing To use End Time/Frame as reception anchors because Metrica has no separate completion flag. They are primary because they are broader and less mechanically convergent than tackles and matchable on possession and ball location. Tackles risk tautological convergence; possession changes mix transition mechanics; zone-entry passes add an arbitrary zone filter. Tackle and precisely event-defined possession-change analyses remain separate frozen sensitivities, not pooled rescue samples.

```python
display(pd.DataFrame([
['Completed open-play reception','Primary','Broad; receiver identifiable','Reception/phase bias'],
['Tackle/challenge','Sensitivity','Precise','Mechanically enriches convergence'],
['Possession change','Sensitivity','Metric-independent','Transition mechanics'],
['Pass into selected zone','Rejected primary','Interpretable','Arbitrary zone definition']
],columns=['Source','Decision','Strength','Risk']))
```

## Controls and matching

Controls are ordinary open-play pseudo-anchors in the same period/possession team, away from receptions and restarts. Primary matching uses only anchor/pre-outcome context. Post-anchor ball displacement, centroid translation, activity, and relational outputs are not primary matching variables because conditioning on them could remove the phenomenon. A sensitivity adds pre-anchor one-second ball and centroid movement bins.

## Leakage and interpretation freeze

Candidate sampling uses event completion, receiver identity, restart status, and availability—not relational outcomes. Control eligibility uses event timing, possession, period, and valid coordinates. Focal selection uses anchor geometry only. Matching uses anchor/pre-anchor context. No post-anchor relational descriptor selects candidates, controls, focal defenders, or matches.

Outcomes A/B/C are frozen in the configuration. Matching failures, unmatched candidates, event dependence, reference reversals, overlap, generic-activity replication, window sensitivity, and historical-example concentration are falsification results, not tuning invitations.

Independent expert annotation is desirable later for semantic validation but is not required for this internal geometric validation.

# Tracking replay prototype

This notebook displays one bounded Metrica Sample Game 2 interval as an observational tracking replay. It makes no causal or tactical claim and requires the local public Metrica files under `data/metrica_sample_game_2/`.

```python
import sys
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
from tracking_animation import animate_tracking_window, to_inline_html
from tracking_animation_metrica_demo import DEMO_SPEC, SOURCE_FPS, load_demo_tracking
```

```python
tracking = load_demo_tracking(ROOT, DEMO_SPEC)
tracking.shape
```

```python
still = animate_tracking_window(
    tracking, DEMO_SPEC, source_fps=SOURCE_FPS, frame_step=2,
    playback_fps=12.5, show_trails=True,
    show_annotations=True, show_rank_highlights=True,
)
anchor_index = min(range(still.displayed_frame_count), key=lambda i: abs(float(still.displayed_frame_ids[i]) - 58401))
still.animation._draw_frame(anchor_index)
still.figure
```

```python
replay = animate_tracking_window(
    tracking, DEMO_SPEC, source_fps=SOURCE_FPS, frame_step=2,
    playback_fps=12.5, trail_seconds=0.4, show_trails=True,
    show_annotations=True, show_rank_highlights=True,
)
to_inline_html(replay)
```

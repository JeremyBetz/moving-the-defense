# Defender-relative movement replay

This notebook reproduces a retrospective analyst replay using public Metrica Sample Game 2 tracking. Each defender's value is accumulated movement relative to the other nine defenders over the trailing two seconds, in metres. The display is descriptive tracking geometry: it does not identify tactical error, defensive quality, marking responsibility, or causal attacker influence. Local Metrica sample-data files are required.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'src'))

from defensive_reorganization_replay_visualization import (
    DEFAULT_SCORE_VMAX_M,
    animate_defensive_reorganization,
    plot_defensive_reorganization_diagnostic,
)
from generate_defensive_reorganization_replay_demo import (
    DISPLAY_SPEC,
    load_approved_scored_demo,
)
from tracking_animation import export_animation, to_inline_html

DISPLAY_CEILING_M = DEFAULT_SCORE_VMAX_M
assert DISPLAY_CEILING_M == 6.25
```

```python
tracking, scores = load_approved_scored_demo(root=ROOT)
# Scores remain in memory; no player-score or tracking table is serialized.
```

```python
diagnostic = plot_defensive_reorganization_diagnostic(
    tracking,
    scores,
    DISPLAY_SPEC,
    selected_time_s=DISPLAY_SPEC.anchor_time_s,
    score_vmax_m=DISPLAY_CEILING_M,
)
diagnostic
```

```python
bundle = animate_defensive_reorganization(
    tracking,
    scores,
    DISPLAY_SPEC,
    frame_step=2,
    playback_fps=12.5,
    trail_seconds=0.4,
    score_vmax_m=DISPLAY_CEILING_M,
    show_team_meter=True,
    show_trails=True,
)
to_inline_html(bundle)

# Optional local export (do not save output in this notebook):
# export_animation(bundle, Path('/tmp/defender_relative_replay_game2.gif'))
```
